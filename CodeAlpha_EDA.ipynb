```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

print(category_sales)
```

```python
plt.figure(figsize=(8,5))

category_sales.plot(kind="bar")

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)

plt.show()
```

```python
region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)

print(region_sales)
```

```python
plt.figure(figsize=(8,5))

region_sales.plot(kind="bar")

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)

plt.show()
```

```python
monthly_sales = df.groupby("Month & Year Order")["Sales"].sum()

print(monthly_sales)
```

```python
df["Month & Year Order"] = pd.to_datetime(df["Month & Year Order"])

monthly_sales = df.groupby("Month & Year Order")["Sales"].sum().sort_index()

plt.figure(figsize=(12,5))

plt.plot(monthly_sales.index, monthly_sales.values)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
top_products = df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10)

print(top_products)
```

```python
plt.figure(figsize=(10,6))

top_products.sort_values().plot(kind="barh")

plt.title("Top 10 Products by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product Name")

plt.tight_layout()
plt.show()
```

# CodeAlpha Exploratory Data Analysis (EDA)

## 📊 Project Overview
This project focuses on performing Exploratory Data Analysis (EDA) on a Superstore sales dataset using Python. The aim is to understand the dataset, identify sales patterns, explore customer behavior, and discover useful business insights through data analysis and visualization.

## 🎯 Objectives
- Understand the structure and characteristics of the dataset.
- Identify top-performing products and customers.
- Analyze sales across regions, states, categories, and customer segments.
- Explore monthly sales trends.
- Present meaningful insights using graphs and charts.

## 🛠️ Technologies Used
- Python
- Pandas
- Matplotlib
- Seaborn
- Google Colab
- Jupyter Notebook

## 🔍 Analysis Performed
- Loaded and explored the Superstore dataset.
- Examined dataset columns and structure.
- Identified the top 10 products by total sales.
- Identified the top 10 customers by total sales.
- Analyzed sales by product sub-category.
- Compared sales across customer segments.
- Analyzed sales by region and state.
- Visualized monthly sales trends.

## 📈 Key Findings
- The Consumer segment generated the highest total sales among the three customer segments.
- The West region recorded the highest total sales among the four regions.
- Bookcases, Chairs, and Phones were among the leading sub-categories by total sales.
- Sales varied over time, with noticeable monthly fluctuations.
- A small group of customers and products contributed substantially to total sales.
